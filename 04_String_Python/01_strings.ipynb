**String Slicing**

```python
s = "Hello World"

print(s)
# Index Start 
print(s[:5])
print((s[0]))
print((s[1]))
print((s[2]))
print((s[3]))
print((s[4]))
print((s[5]))
print((s[6]))
print((s[7]))
print((s[8]))
print((s[9]))
print((s[10]))

# String start from 0
print(s[:10])

# Step slicing 
print(s[0:10:2])

# slicing end 
print(s[5:])

# Slicing is done throgh square brackets = []

# Negetive Slicing 
print(s[-1:])
print(s[-2:])
print(s[-3:])
print(s[-4:])
print(s[-5:])
print(s[-6:])
print(s[-7:])
print(s[-8:])
print(s[-9:])
print(s[-10:])
print(s[-11:])

print(s[-11:-1:2])

```

## Strings Methods 

**capitalize()**

**Syntax - string.capitalize()**

```python
text = "hello world"
print(text.capitalize())
```

```python
text = "PYTHON PROGRAMMING"
print(text.capitalize())
```

```python
text = "hello welcome to the world of the programming"
print(text.capitalize())
```

**center()**

**Syntax - string.center(width, fillchar)**

```python
text = "Python"
print(text.center(20))
```

```python
text = "Python"
print(text.center(20,"*"))
```

**Count()**

**Syntax - string.count(substring, start, end)**

```python
text = "python is easy, python is a powerful and python is a high level language"

print(text.count("python"))
print(text.count("is"))
```

```python
text = "banananananananananananana"
print(text.count("a",2,13))
print(text.count("n",11))
```
