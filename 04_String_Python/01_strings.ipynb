**Declaration of Strings**

```python
# Single Quotes 
single_quotes = 'uzmaki naruto'
print(single_quotes)

# Double Quotes 
double_quotes = "Sheyo Hinata"
print(double_quotes)

# Triple Single Quotes 
triple_single_quotes = '''Akaza'''
print(triple_single_quotes)

# Triple Double Quotes 
triple_double_quotes = """Rengoku"""
print(triple_double_quotes)
```

**String Slicing**

```python
s = "Hello World"

print(s)
# Index Start 
print(s[:5])
print((s[0]))
print((s[1]))
print((s[2]))
print((s[3]))
print((s[4]))
print((s[5]))
print((s[6]))
print((s[7]))
print((s[8]))
print((s[9]))
print((s[10]))

# String start from 0
print(s[:10])

# Step slicing 
print(s[0:10:2])

# slicing end 
print(s[5:])

# Slicing is done throgh square brackets = []

# Negetive Slicing 
print(s[-1:])
print(s[-2:])
print(s[-3:])
print(s[-4:])
print(s[-5:])
print(s[-6:])
print(s[-7:])
print(s[-8:])
print(s[-9:])
print(s[-10:])
print(s[-11:])

print(s[-11:-1:2])

```

**String Formatting operators**

```python
name = "Suraj"
age = 20

print("my name is " + name + " and I am" + str(age) + " years old.")
```

```python
## String Formatting 
name = "Suraj"
print("My name is %s" %name)
```

```python
## integer Formatting 
age = 20 
print("I an %d years old " %age)
```

```python
## Float Formatting 
price = 99.4567
print("Price is Petrol is %f" %price)
```

```python
## Multiple Values 

name = "Suraj"
marks = 95
print("My name is %s and I Scored %d marks " %(name, marks))
```
