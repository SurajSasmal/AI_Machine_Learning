**Type Casting** : Type Casting means converting this one datatype to another datatype String -> Integer -> String -> Floating -> String

**Implict Type Casting**

```python
x = 10
y = 2.5 

z = x + y
print(z)
```

**Explict Type Casting**

```python
x = 3.5
print(x,type(x))

y = int(x)
print(y,type(y))
```

**Object reference**

```python
a = 10 # -> 10 is an object and a is a reference pointing to the object
b = a # ->b is also a new reference pointing to the same object 

print(id(a))
print(id(b))

print(id(a) == id(b))
```

```python

# Swapping two variables 

a = 5
b = 6

print("Before Swapping : ")
print(f"a : {a}")
print(f"b : {b}")

b = a + b
a = b - a
b = b - a

print("After Swapping : ")
print(f"a : {a}")
print(f"b : {b}")
```
