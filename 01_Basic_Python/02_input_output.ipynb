**Input and Output Function**

```python
name = input("Enter your name : ") # <- input function
print(name)  # <- Output function
```

Taking multiple input using **split()** method 

```python

x, y, z = input("Enter the value : ").split()

print(x, y, z)
```

Changing the type of input 

```python
age = int(input("Enter the age of the person : "))
print(age)
```

```python

height = float(input("Enter the height of the person : "))
print(height)
```

```python
name = input("Enter the name of the person : ")
print(name)
```

**Type Function**

```python

age = 67
print(age,type(age))
```

```python

height = 5.6
print(height,type(height))
```

```python

name = "Suraj Sasmal"
print(name,type(name))
```

```python

fruits = ["Apple", "Banana", "Guava", "Pineapple"]
print(fruits, type(fruits))
```

**isinstance()**

```python
x = 123
print(isinstance(x,int))
print(isinstance(x,float))
```

```python
name = "Suraj"
print(isinstance(name,str))
```
