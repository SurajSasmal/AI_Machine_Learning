```python
# Assign 

a = 34
print(a)
```

```python
# Add and Assign 
a = 45
a += 23
print(a)
```

```python
# Subtract and Assign 
a = 23
a -= 10
print(a)
```

```python
# Multiple and Assign 
a = 45
a *= 3
print(a)
```

```python
# Left Shift and Assign 

a = 10
a <<= 2

print(a)
```

```python
# Right Shift and Assign 


a = 12

a >>= 1
print(a)
```

```python
# Divide and Assign 
a = 12
a /= 2

print(a)
```

```python
# Floor divide and assign 
a = 34
a //= 5
print(a)
```
