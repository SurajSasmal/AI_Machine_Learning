```python
# Equals to 
a = 10
b = 10

print(a == b)
print(a != b)
```

```python
# Not Equals to 
a = 10
b = 15
print(a != b)
print(a == b)
```

```python
# Greater than 
a = 20
b = 5
print(a > b)
print(a < b)
```

```python
# Less than 
a = 20
b = 50
print(a < b)
print(a > b)
```

```python
# Greater than equal to 

a = 20
b = 19
print(a >= b)
```

```python
# less than equal to 
a = 19
b = 20
print(a <= b)
```
