```python
# Bitwise AND 

a = 5 
b = 3

print(a & b)
```

```python
# BitWise OR 
a = 5 
b = 3
print(a | b)
```

```python
# Bitwise XOR 
a = 5
b = 3
print(a ^ b)
```

```python
# Bitwise NOT 
print(~a)
```

```python
# Left Shift
a = 5
b = 3

print(a << b)
```

```python
# Right Shift 
a = 5 
b = 1


print(a >> b)
```
