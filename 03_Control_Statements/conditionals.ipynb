## If Statement 

```python
age = 20

if age >= 18:
    print("Eligible to Vote.")
```

```python
age = int(input("Enter the age of the person : "))

if age >= 18:
    print("You can apply for driving license")
```

## IF Else Statement

```python
age = 10

if age <= 12:
    print("Travel for free.")
else:
    print("Pay 10$ for ticket.")
```

```python
age = int(input("Enter the age of the person : "))

if age >= 18:
    print("You can apply for driving license")
else:
    print(f"You can'nt apply for driving license.\nYou have to wait for {18 - age} years.")
```

**Short hand if else**

```python
marks = int(input("Enter the mark of the student : "))
res = "Pass" if marks >= 40 else "Fail"
print(res)
```

## elif Statement 

```python
age =int(input("Enter the age of the person : "))

if age <= 12:
    print("Child.")
elif age <= 19:
    print("Teenager.")
elif age <= 35:
    print("Young Adult.")
else:
    print("Adult")
```

## Nested If Else 

```python
age = int(input("Enter the age of the person : "))

is_member = input("Is the Person is a member or not(Yes/No): ").lower()\

if age >= 60:
    if is_member == "yes":
        print("30% Senior discount.")
    else:
        print("20% Senior discount.")
else:
    print("Not eligible for a senior discount.")
```

## match Case

```python
number = 3

match number:
    case 1:
        print("One")
    case 2:
        print("Two")
    case 3:
        print("Three")
    case _:
        print("default")
```
